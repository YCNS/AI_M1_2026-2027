```python
# I - Data cleaning
```

```python
# Check for duplicates and removing them
```

```python
import pandas as pd

dflab2 = pd.read_csv('students.csv')

duplicates = dflab2.duplicated() # Boolean array 

# Number of duplicate rows
print(dflab2[duplicates]) 
print(f"Number of duplicate rows: {duplicates.sum()}")  
```

```python
dflab2_cleaned = dflab2.drop_duplicates() # Dropping duplicates and storing cleaned dataframe into dflab2_cleaned
dflab2_cleaned 
```

```python
# Number of duplicate rows 
print(f"Number of duplicate rows: {dflab2_cleaned.duplicated().sum()}") 
```

```python
# Dropping duplicates directly from the initial dataframe
dflab2.drop_duplicates(inplace=True)
dflab2
```

```python
# Number of duplicate rows 
print(f"Number of duplicate rows: {dflab2.duplicated().sum()}") 
```

```python
# Checking for missing data and processing them 
```

```python
missing_values = dflab2.isnull().sum() 
print(missing_values) 
```

```python
# Handling missing data
```

```python
# Remove missing data:
dflab2= dflab2.dropna() 
print(dflab2.isnull().sum()) 
```

```python
# Imputation
dflab2['Score'] = dflab2['Score'].fillna(dflab2['Score'].mean())
print(dflab2.isnull().sum()) 
```

```python
# II - Data normalization  
```

```python
from sklearn.preprocessing import MinMaxScaler 

scaler = MinMaxScaler() 
dflab2['Age'] = scaler.fit_transform(dflab2[['Age']]) 

print(dflab2) 
```

```python
# III - Data encoding
```

```python
from sklearn.preprocessing import LabelEncoder 
```

```python
# Label Encoding
encoder = LabelEncoder() 
dflab2['Gender'] = encoder.fit_transform(dflab2['Gender']) 

print(dflab2.head())
```

```python
# One-Hot Encoding
df_one_hot = pd.get_dummies(dflab2, columns=['City'], drop_first=True) 
print(df_one_hot)
```
