```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(-10, 10.1, 0.1)
n = len(x) ; 
y = np.arange(-10, 10.1, 0.1)
```

```python
def activationFunc(fct, x, y, n):
    match fct:
         case 'Step':
            for i in range(n):
                if(x[i]<0):
                    y[i]=0 
                else:
                    y[i]=1 
            plt.plot(x, y)
            plt.show()
         case 'Linear':
              y=x
              plt.plot(x, y)
              plt.show()
         case 'ReLU':
             y=x
             for i in range(n):
                if(x[i]<0):
                   y[i]=0
             plt.plot(x, y)
             plt.show()
         case 'Sigmoid':
            y=1/(1+np.exp(-x))
            plt.plot(x, y)
            plt.show()
         case 'TanH':
            y=(np.exp(x) - np.exp(-x))/(np.exp(x) + np.exp(-x)) 
            plt.plot(x, y)
            plt.show()
         case _:
            print('Not reconized function')
```

```python
fct = input('Input the activation function. Example : Step, Linear, ReLU, Sigmoid, TanH : s')
activationFunc(fct, x, y, n)
```

```python
# Input Vector P 
P = np.array([3 ,-2, 0, -1.5])

# Weight Vector W 
W = np.array([1, 2, 3, 4])

# Bias b  
b = 1 ; 

# Computing the Potential n 
n = W*P+b 

# Formal neuron output a 
a = (np.exp(n)-np.exp(-n))/(np.exp(n)+np.exp(-n)) 
print(a)
```
