```python
# Setting modules
import os
import matplotlib.pyplot as plt

os.chdir("..")
print(os.getcwd())

from utils.helpers import clean_lines, parse_products
from utils.product import Product
```

```python
# Reading the raw data 
with open("data/raw/data.txt", "r") as f:
    raw_lines = f.readlines()

products = [Product(n, p, q) for n, p, q in parse_products(clean_lines(raw_lines))]

names  = [p.name for p in products]
totals = [p.total_value() for p in products]
print(dict(zip(names, totals)))
```

```python
# The visualization
plt.bar(names, totals, color="steelblue")
plt.xlabel("Product")
plt.ylabel("Total sales")
plt.title("Total sales per product")
plt.show()
```
