# Student data simulation and analysis

```python
import pandas as pd
from students import generate_students
from analysis import add_ages, compute_statistics, save_report
```

## 1. Generate the data

```python
students = generate_students()   # add seed=42 for reproducible results
students = add_ages(students)

df = pd.DataFrame(students)
df
```

## 2. Statistical analysis

```python
stats = compute_statistics(students)
for key, value in stats.items():
    print(f"{key:20s}: {value:.2f}" if isinstance(value, float) else f"{key:20s}: {value}")
```

## 3. Save the report

```python
path = save_report(students, stats)   # -> results/report.txt
print("Report saved to:", path)
print()
print(open(path, encoding="utf-8").read())
```
